```python
samp1 = 77
cal1 = (samp1 - 32) * 5 / 9
print(cal1)

samp2 = 90
cal2 = (samp2 - 32) * 5 / 9
print(cal2)

cal3 = cal2 - cal1
print(cal3)
```

```python
def farenheit_to_cal(farenheit):
  return (farenheit - 32) * 5 / 9
print(farenheit_to_cal(90))
print(farenheit_to_cal(20))
print(farenheit_to_cal(46))
```
